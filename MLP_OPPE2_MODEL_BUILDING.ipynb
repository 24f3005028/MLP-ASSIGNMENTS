##Q2 [Marks =3] Break the dataset into features(`X`) and label (`y`), where the column `satisfaction` goes to `y` and the rest of the columns go to `X`. How many data points belonging to `satisfaction` value 1 are there in the dataset ? [NAT]


Sol: In supervised learning, you separate inputs (features) and output (label). Here, satisfaction is the binary label (0/1), and all other columns are features.

To answer the question, you simply count how many rows in y are equal to 1 using value_counts or a boolean mask

```python
X = mb_df.drop(columns=['satisfaction'])
Y = mb_df['satisfaction']

Y.value_counts()
```

##Q3 [Marks : 3] Split the dataset into train and test dataset into the `80:20` ratio while keeping `random_state =64`. What is the shape of the training label vector(`y_train`) dataset?

(A) (20000,)

(B) (16574,27)

(C) (20024,14)

(D) (4522,98)

(E) (16574,)

##Sol:
 train_test_split is used to evaluate models: training data is used to fit, and test data is used to measure performance on unseen examples.

OPPE always fixes test_size=0.2 and a specific random_state so your answer is reproducible; you then inspect y_train.shape.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, random_state=64, test_size=0.2
)
```

```python
print("Y_train shape:", Y_train.shape)
```

##Q4 [4 marks][NAT] Take `LogisticRegression` estimator with following parameters for training:
  - Use `sag` as solver
  - Set random state to be equal to 64
  - Tolerance for stopping criteria to be 1e-3
  - Maximum number of iterations taken for the solvers to converge to be 100

Enter the recall score for the given model using test set(X_test, y_test)


##Sol:
Logistic regression is a standard linear classifier for binary labels like satisfaction.

Recall measures how many of the positive samples (here label 1) you correctly identify; OPPE uses it to check if you can compute different metrics, not just accuracy.

You fit on training data, predict on test data, and compute recall_score(y_test, y_pred). By default, pos_label=1.

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(
    solver='sag',
    random_state=64,
    tol=1e-3,
    max_iter=100
)
```

```python
model.fit(X_train, Y_train)
pred = model.predict(X_test)
```

```python
from sklearn.metrics import recall_score
recall_sr = recall_score(Y_test, pred)
print("Logistic recall:", recall_sr)
```


## Q8. [4 marks] [MCQ] Use Gridsearchcv with KNeighborsClassifier estimator, accuracy as scoring parameter, cv= 5. Consider [1,3,7,11] as K values to be examined.
* Consider following parameters for KNeighborsClassifier:

 - Take metric as 'minkowski',
 - Set P value as 2

- Keep other parameter values as default values.

What is the best value of K you obtained using the above instructions?

Options

A) 1

B) 3

C) 7

D) 11

Ans:

V1: C


## Sol:
Use KNeighborsClassifier + GridSearchCV:

metric='minkowski', p=2 (Euclidean distance),

n_neighbors in [1, 3, 7, 11],

scoring='accuracy', cv=5.
Report best k.

Idea and approach
KNN classifies a point by the majority label among its nearest neighbours; n_neighbors controls bias–variance trade‑off.

GridSearchCV tests each n_neighbors across 5 folds of training data, scoring by accuracy; the best hyperparameters minimize cross‑validated error

```python
from sklearn.neighbors import KNeighborsClassifier
model4 = KNeighborsClassifier(
    metric="minkowski",
    p=2
)

```

```python
from sklearn.model_selection import GridSearchCV

param_grid={
    "n_neighbors":[1,3,7,11]
}
tuned_m4 = GridSearchCV(estimator=model4, param_grid=param_grid, scoring="accuracy",cv=5)
```

```python
tuned_m4.fit(X_train, Y_train)
#m4_pred = tuned_m4.predict(X_test)
tuned_m4.best_params_
```
