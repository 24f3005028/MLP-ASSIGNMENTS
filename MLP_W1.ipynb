This notebook contains the solutions for the graded dataset assignment.

### 1. Setup and Data Loading

```python
import pandas as pd
import gdown

# Download and load dataset
file_id = '1x-phOl0d1QHy4ea8O_hHdz_wBGH35dRz'
url = f'https://drive.google.com/uc?id={file_id}&export=download'
gdown.download(url, 'dataset.csv', quiet=True)

df = pd.read_csv('dataset.csv')
df.replace('?', pd.NA, inplace=True)
print(f"Dataset loaded with {df.shape[0]} rows and {df.shape[1]} columns.")
```

### 2. Data Inspection

```python
display(df.head())
print("Columns:", df.columns.tolist())
```

### 3. Basic Row and Column Selection

```python
# Q1 & Q2: Even indexed rows
q1_df = df.iloc[::2]
ans1 = q1_df.iloc[0, 3]
ans2 = q1_df.iloc[332, 3]

# Q3: Odd indexed rows
q3_df = df.iloc[1::2]
ans3 = q3_df.iloc[332, 3]

# Q4: Odd indexed columns
q4_df = df.iloc[:, 1::2]
ans4 = q4_df.iloc[100, 5]

# Q5: Even rows and columns
q5_df = df.iloc[::2, ::2]
ans5 = q5_df.iloc[255, 3]

print(f"Q1: {ans1}, Q2: {ans2}, Q3: {ans3}, Q4: {ans4}, Q5: {ans5}")
```

### 4. Advanced Filtering and Aggregation

```python
# Q6: Recent 6 years
recent_yrs = sorted(df['Year'].unique())[-6:]
ans6 = df[df['Year'].isin(recent_yrs)].shape[0]

# Q7 & Q8: Rooms and Bathrooms
ans7 = df[(df['num_rooms'] == 3) & (df['num_bathrooms'] == 3)].shape[0]
ans8 = df[(df['num_rooms'] == 3) | (df['num_bathrooms'] == 3)].shape[0]

# Q9: Specific 2022 Greenwich criteria
ans9 = df[(df['Year'] == 2022) & (df['Locality'] == 'Greenwich') & (df['num_rooms'] == 3) & (df['Face'].isin(['North', 'East']))].shape[0]

# Q10: August sales
ans10 = df[pd.to_datetime(df['Date']).dt.month == 8].shape[0]

# Q11: Highest average price locality
ans11 = df.groupby('Locality')['Sale Price'].mean().idxmax()

print(f"Q6: {ans6}, Q7: {ans7}, Q8: {ans8}, Q9: {ans9}, Q10: {ans10}, Q11: {ans11}")
```

### 5. Summary Table

```python
results = {
    "Question": ["Q1", "Q2", "Q3", "Q4", "Q5", "Q6", "Q7", "Q8", "Q9", "Q10", "Q11"],
    "Result": [ans1, ans2, ans3, ans4, ans5, ans6, ans7, ans8, ans9, ans10, ans11]
}
display(pd.DataFrame(results))
```

## Question 1

**Task:** Select all even indexed rows and find the value in the 3rd indexed column of the 0th indexed row.

```python
q1_df = df.iloc[::2]
q1_answer = q1_df.iloc[0, 3]
print("Question 1 Answer:", q1_answer)
```

## Question 2

**Task:** Select all even indexed rows and find the value in the 3rd indexed column of the 332nd indexed row.

```python
q2_answer = q1_df.iloc[332, 3]
print("Question 2 Answer:", q2_answer)
```

## Question 3

**Task:** Select all odd indexed rows and find the value in the 3rd indexed column of the 332nd indexed row.

```python
q3_df = df.iloc[1::2]
q3_answer = q3_df.iloc[332, 3]
print("Question 3 Answer:", q3_answer)
```

## Question 4

**Task:** Select all odd indexed columns and find the value in the 5th indexed column of the 100th indexed row.

```python
q4_df = df.iloc[:, 1::2]
q4_answer = q4_df.iloc[100, 5]
print("Question 4 Answer:", q4_answer)
```

## Question 5

**Task:** Select all even indexed rows and even indexed columns, then find the value in the 3rd indexed column of the 255th indexed row.

```python
q5_df = df.iloc[::2, ::2]
q5_answer = q5_df.iloc[255, 3]
print("Question 5 Answer:", q5_answer)
```

## Question 6

**Task:** Determine the number of samples belonging to the six most recent years.

```python
recent_years = sorted(df['Year'].unique())[-6:]
q6_answer = df[df['Year'].isin(recent_years)].shape[0]
print("Question 6 Answer:", q6_answer)
```

## Question 7

**Task:** Determine the number of samples where `num_rooms = 3` and `num_bathrooms = 3`.

```python
q7_answer = df[(df['num_rooms'] == 3) & (df['num_bathrooms'] == 3)].shape[0]
print("Question 7 Answer:", q7_answer)
```

## Question 8

**Task:** Determine the number of samples where `num_rooms = 3` or `num_bathrooms = 3`.

```python
q8_answer = df[(df['num_rooms'] == 3) | (df['num_bathrooms'] == 3)].shape[0]
print("Question 8 Answer:", q8_answer)
```

## Question 9

**Task:** In 2022, count the number of houses in Greenwich with `num_rooms = 3` and facing either North or East.

```python
q9_answer = df[
    (df['Year'] == 2022) &
    (df['Locality'] == 'Greenwich') &
    (df['num_rooms'] == 3) &
    (df['Face'].isin(['North', 'East']))
].shape[0]
print("Question 9 Answer:", q9_answer)
```

## Question 10

**Task:** Count the number of samples corresponding to the month of August across all years.

```python
q10_answer = df[pd.to_datetime(df['Date']).dt.month == 8].shape[0]
print("Question 10 Answer:", q10_answer)
```

## Question 11

**Task:** Identify the locality with the highest average Sale Price.

```python
q11_answer = df.groupby('Locality')['Sale Price'].mean().idxmax()
print("Question 11 Answer:", q11_answer)
```

## Final Answers Summary

```python
answers = {
    "Q1": q1_answer, "Q2": q2_answer, "Q3": q3_answer, "Q4": q4_answer, "Q5": q5_answer,
    "Q6": q6_answer, "Q7": q7_answer, "Q8": q8_answer, "Q9": q9_answer, "Q10": q10_answer,
    "Q11": q11_answer
}
summary_df = pd.DataFrame(list(answers.items()), columns=["Question", "Answer"])
display(summary_df)
```
