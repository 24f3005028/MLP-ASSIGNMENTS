Answer: B

```python
l=df.columns.to_list()
l
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline


si=SimpleImputer(strategy="most_frequent")

pipe = Pipeline(
    [("imputer",si)]
).set_output(transform="pandas")

df[l]=pipe.fit_transform(df[l])
```
